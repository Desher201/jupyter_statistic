```python
import pandas as pd
from requests.packages import package

df = pd.read_csv(r"E:\PythonProject\jupyter_statistic\data\robot_runs.csv",delimiter=",")
print(df.head())
print(len(df))
print(df.columns)
print(df.dtypes)
```

`battery_before_percent` - заряд акумулятора.

`route_time_sec` - час проходження маршруту.           <---------- три випадкові величини

`package_weight_kg` - вага вантажу.

```python
df_avg = df[[
    "battery_before_percent",
    "route_time_sec",
    "package_weight_kg"
]].mean()
print(df_avg)
```

```python
df_median = df[[
    "battery_before_percent",
    "route_time_sec",
    "package_weight_kg"
]].median()
print(df_median)
```

```python
df_mode = df[[
    "battery_before_percent",
    "route_time_sec",
    "package_weight_kg"
]].mode()
print(df_mode)
```

```python
df_var = df[[
    "battery_before_percent",
    "route_time_sec",
    "package_weight_kg"
]].var(ddof=0)
print(df_var)
```

```python
df_var = df[[
    "battery_before_percent",
    "route_time_sec",
    "package_weight_kg"
]].var(ddof=0)
print(df_var)
```

```python
std_route_time = df[[
    "battery_before_percent",
    "route_time_sec",
    "package_weight_kg"
]].std(ddof=0)
print(std_route_time)
```

```python
sample = df[[
    "battery_before_percent",
    "route_time_sec",
    "package_weight_kg"
]].sample(n=10, random_state=42)
print(sample)
```

```python
sample_avg = df[[
    "battery_before_percent",
    "route_time_sec",
    "package_weight_kg"
]].sample(n=10, random_state=42).mean()
print(sample_avg)
```

```python
sample_var = df[[
    "battery_before_percent",
    "route_time_sec",
    "package_weight_kg"
]].sample(n=10, random_state=42).var(ddof=1)
print(sample_var)
```

```python
sample_var = df[[
    "battery_before_percent",
    "route_time_sec",
    "package_weight_kg"
]].sample(n=10, random_state=42).std(ddof=1)
print(sample_var)
```

```python
final = pd.DataFrame({
    "Показник": [
        "Середній час маршруту",
        "Дисперсія часу",
        "Стандартне відхилення часу",
        "Середня вага",
        "Дисперсія ваги",
        "Стандартне відхилення ваги"
    ],
    "Генеральна сукупність": [
        df["route_time_sec"].mean(),
        df["route_time_sec"].var(ddof=0),
        df["route_time_sec"].std(ddof=0),
        df["package_weight_kg"].mean(),
        df["package_weight_kg"].var(ddof=0),
        df["package_weight_kg"].std(ddof=0)
    ],
    "Вибірка": [
        sample["route_time_sec"].mean(),
        sample["route_time_sec"].var(ddof=1),
        sample["route_time_sec"].std(ddof=1),
        sample["package_weight_kg"].mean(),
        sample["package_weight_kg"].var(ddof=1),
        sample["package_weight_kg"].std(ddof=1)
    ]
})
print(final)
```
